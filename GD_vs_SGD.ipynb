IRIS DATASET

Comparison of Gradient Descent
and
Stochastic Gradient Descent

Neural Network: # Input Layer: 4 neurons # Hidden Layer: 10 neurons # Hidden Layer2: 8 neurons # Output Layer: 3 neurons


```python
import numpy as np
import tensorflow as tf
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

 Load Iris dataset

```python
iris = load_iris()
X = iris.data
y = iris.target
print("Total samples:", X.shape[0])
print("Number of input features:", X.shape[1])
print("Number of classes:", len(np.unique(y)))
```

Standardize the input features

```python
scaler = StandardScaler()
X = scaler.fit_transform(X)
```

Divide dataset into training and testing

```python
X_train, X_test, y_train, y_test = train_test_split(
X,
y,
test_size=0.20,
random_state=42,
stratify=y
)
print("\nTraining samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

Create Neural Network

```python
def create_model():
  model = Sequential()
  model.add(
      Dense(
          10,
          activation='relu',
          input_shape=(4,)
      )
  )
  # Hidden Layer 2
  model.add(
      Dense(
          8,
          activation='relu'
          )
      )
  # Output Layer
  model.add(
      Dense(
          3,
          activation='softmax'
          )
      )
  return model
```

GRADIENT DESCENT

```python
print("GRADIENT DESCENT")
model_gd = create_model()
model_gd.compile(
    optimizer=tf.keras.optimizers.SGD(
          learning_rate=0.01
           ),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
print("Batch size:", len(X_train))
print("Epochs: 50")
history_gd = model_gd.fit(
    X_train,
    y_train,
    batch_size=len(X_train),
    epochs=50,
    verbose=1
    )
```

Evaluate Gradient Descent model

```python
gd_loss, gd_accuracy = model_gd.evaluate(
    X_test,
    y_test,
    verbose=0
    )
print("\nGradient Descent Test Loss:", gd_loss)
print("Gradient Descent Test Accuracy:", gd_accuracy)
```

STOCHASTIC GRADIENT DESCENT

```python
print("STOCHASTIC GRADIENT DESCENT")
model_sgd = create_model()
model_sgd.compile(
    optimizer=tf.keras.optimizers.SGD(
        learning_rate=0.01
        ),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
    )
# batch_size = 1
 # means one training sample is used
 # for each weight update.
print("Batch size: 1")
print("Epochs: 50")
history_sgd = model_sgd.fit(
    X_train,
    y_train,
    batch_size=1,
    epochs=50,
    shuffle=True,
    verbose=1
    )
# Evaluate SGD model
sgd_loss, sgd_accuracy = model_sgd.evaluate(
    X_test,
    y_test,
    verbose=0
    )
print("\nStochastic Gradient Descent Test Loss:", sgd_loss)
print("Stochastic Gradient Descent Test Accuracy:", sgd_accuracy)
```

FINAL COMPARISON

```python
print("FINAL COMPARISON")
print("\nGradient Descent")
print("Test Loss     :", gd_loss)
print("Test Accuracy :", gd_accuracy)
print("\nStochastic Gradient Descent")
print("Test Loss     :", sgd_loss)
print("Test Accuracy :", sgd_accuracy)
```
